```python
df = spark.read.csv(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee.csv"
                    ,header = True
                    ,inferSchema = True
                    ,sep = "|"
                    ,quote ="'"
                    
                    )
df.display()
```

##SQL View 
>> It is virtual tables  that can created from DF

>>Create Temp View

```python
df.select("id","name").display()
df.createOrReplaceTempView("employee_tem_view")    
```

>>Accessing Temp View

```python
%sql

SELECT * FROM employee_tem_view;
```

##Display all Tem Views

```python
for view in spark.catalog.listTables():
    print(view)
    
```

##Global Temp Table is accessible accross all spark

```python
df.createGlobalTempView("employee_global_view")
```

```python
for view in spark.catalog.listTables("global_temp"):
    if len(view.namespace) == 0:
        print(f"Temporary View: {view.name}")
    else:
        print(f"Global view: {view.name}")
```
