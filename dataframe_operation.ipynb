## Prerequisites

```python
df = spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee.csv",
    header=True,
    inferSchema=True,
    sep="|",
    quote="'"
)
df.display()
```

## SELECT COLUMN

## Using (STR) method

```python
df.select("id","name").display()
```

##Using(COL) method

```python
from pyspark.sql.functions import col
df.select(col("id").alias("employee_id"),col("name").alias("employee_name")).display()
```

##Exclude Columns

```python
column_to_exclude =["id"]

column_to_include = [c for c in df.columns if c not in column_to_exclude]

# for i in df.columns:
#     if i not in column_to_exclude:
#         column_to_include.append(i)
      
df.select(column_to_include).display()
#df.select(*column_to_include).display()
```

## Filter rows with filter 

```python
from pyspark.sql.functions import col
df.filter((col("desig")=="Developer") | (col("desig")=="Team Lead")).display()
```

## Problem - Fetch only the name of the employes

```python
df.select(col("name")).filter(col("gen")=="M").display()
```

```python
df.filter(col("company")=='Cisco').display()
```

```python
from pyspark.sql.functions import lower

df.filter(lower("company")=='cisco').display()
```

##Grouping Aggregation

```python
#df.groupBy("gen").count().sort(col("count"),ascending=True).display()
df.groupBy("gen").count().sort(col("count").desc()).display()
```

##Derive New Columns

```python
from pyspark.sql.functions import lit

df.withColumn("is_alive", lit("India")).display()
```

```python
from pyspark.sql.functions import when

df.withColumn("category",
              when(col("exp")>=10,"senior").
              when(col("exp")>=5,"Mid Level").
              when(col("exp")>=0,"Junior").
              otherwise("Intern")
              
              ).limit(4).display()

```

```python
df.selectExpr(
"*",
"""
CASE
    WHEN exp>=10 THEN 'Senior'
    WHEN exp>=5  THEN 'MID Level'
    WHEN exp>=0  THEN 'Junior'
    ELSE 'Intern'
END as category

"""


).display()
```
