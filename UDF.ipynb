##Prerequisits 

```python
df = spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee.csv",
    header=True,
    inferSchema=True,
    sep="|",
    quote="'"
)
df.display()
```

##Apply python function in pyspark

>>Define Python Function

```python
def convert_to_title_case(st):
  if st:
    return st.title()
  else:
    return None

```

```python
df.select(convert_to_title_case("name")).display()
```

##Convert python function in UDF

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

convert_to_title_case_udf = udf(convert_to_title_case,StringType())

df.select(convert_to_title_case_udf("name").alias("name")).display()
```

##Pyspark UDF in SQl

```python
df.createOrReplaceTempView("employee_udf_sql")
```

>>register udf

```python
spark.udf.register("convert_to_title_case_udf_sql",convert_to_title_case,StringType())
```

```python
%sql

SELECT convert_to_title_case_udf_sql(name) FROM employee_udf_sql;
```
